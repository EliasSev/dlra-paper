```python
from dlra.solvers import DirectSolver
from dlra.problem import TestProblemSetup
from dlra.viz import plot_source

from fenics import Function
import numpy as np
from numpy.typing import NDArray
```

```python
pb = TestProblemSetup(n=64).build(name='V')

plot_source([pb.weights, pb.weights_star], pb.Vh)
```

```python
solver = DirectSolver(pb.operator)

x1 = solver.solve(y=pb.y, w=pb.weights, lambda_=1e-4)
x2 = solver.solve(y=pb.y, w=pb.weights_star, lambda_=1e-4)

plot_source([x1, x2], pb.Vh)
```
